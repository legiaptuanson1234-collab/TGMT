```python
import os
import glob

# BẠN HÃY SỬA LẠI ĐƯỜNG DẪN NÀY CHO ĐÚNG VỚI MÁY CỦA BẠN NHÉ
# Trỏ vào thư mục xebagac mà bạn vừa giải nén
path_to_data = r'D:\TGMT\TGMTNC\xebagac' 

txt_files = glob.glob(os.path.join(path_to_data, '**', 'labels', '*.txt'), recursive=True)

count = 0
for file_path in txt_files:
    with open(file_path, 'r') as f:
        lines = f.readlines()
    
    with open(file_path, 'w') as f:
        for line in lines:
            parts = line.split()
            # Nếu nhãn đang là 0 thì đổi thành 4
            if len(parts) > 0 and parts[0] == '0':
                parts[0] = '4' 
                f.write(" ".join(parts) + "\n")
            else:
                f.write(line)
    count += 1

print(f"Thành công! Đã xử lý {count} file nhãn sang class 4 (xe ba gác).")
```
